## Shapefiles to GeoPackage

This is a data cleaning notebook for my own convenience. I'm tired of try to share a dozen shapefiles with collaborators. 

```python
from pathlib import Path
import geopandas as gpd
import sys
```

```python
input_dir = Path(r"C:\Users\cleor\Documents\ArcGIS\Projects\TrendRasterHotspotAnalysis\outputs\aggregation\oct_2026_aggregated")
output_gpkg = input_dir / "aggregated_oct_2026.gpkg"
```

```python
shapefiles = sorted(input_dir.rglob("*.shp"))
```

```python
# export each shapefile
for shp in shapefiles:
    layer_name = shp.stem

    print(f"Adding {layer_name}...")

    gdf = gpd.read_file(shp)

    gdf.to_file(
        output_gpkg,
        layer=layer_name,
        driver="GPKG"
    )

    print(f"Exported: {layer_name}")

# Verify
print(gpd.list_layers(output_gpkg))
```
